## 1. The Three Places You Will Work

```text
GitHub.com
remote copy of your repository
        ↓ clone
GitHub Desktop
manages the connection between local and remote
        ↓ open
VS Code
where you edit notebooks, Markdown, Python, and other files
```

After you make changes:

```text
VS Code
edit + save
    ↓
GitHub Desktop
review + commit
    ↓
Push
    ↓
GitHub.com
updated remote copy
```

### Important vocabulary

**Remote repository** — the copy stored on GitHub.com.  
**Local repository** — the copy stored on your computer.  
**Clone** — create a local copy of a GitHub repository.  
**Commit** — record a set of changes with a description.  
**Push** — send committed local changes to GitHub.


### Important: Your Repository Should Have One Home

When you clone your GEOG 592 repository, choose **one permanent location on your computer** and continue working from that same folder.

Avoid having multiple copies in different folders (such as on your desktop, download, and documents folder.)

Instead, you should put the folder somewhere obvious, like 

```text
Documents/
└── dev/
    └── geog592-abc123/
```

GitHub Desktop and VS Code should both use **that same folder**.

The point is that there should be just *one repository on your computer*, and *one repository on Github.*

## 4. Preserve the Repository Structure

Your repository has a folder/subfolder structure which is preserved on your computer and on Github. 
Example:

```text
geog592-jarce/
│
├── README.md
├── schedule/
│   ├── week01/
│   ├── week02/
│   └── ...
├── homework/
│   ├── hw01/
│   └── hw02/
├── project/
└── resources/
```

This structure is important.  If you are asked to put a notebook in a particular folder, you should be sure to keep it there, since later programs might need to find it.  Moving or renaming a file may cause errors in code that uses the file.


## 8. Make and Save a Change in VS Code

For practice:

1. Open `README.md`.
2. Make the requested change.
3. Save the file.

You may also edit a notebook or Python file.

### Saving is not committing

Saving will only change the file on your computer.  A Git repository also includes a history of each change to a file, **but the Git history is not modified when you save the file, and nothing changes on Github when you save your file locally.**

You update the local history when you **commit** your changes using the Git command.

You upload your updated history and updated file(s) to Github when you **push** your commit.

```text
edit → save → commit → push
```


## 9. Review the Change in GitHub Desktop

Return to GitHub Desktop.

The edited file should appear under **Changes**. A modified file may appear with an `M`.

Before committing, check to make sure the change you made is the change you wanted (you edited the correct file, and didn't delete or move anything else.)

## 10. Commit the Change

A commit records a meaningful checkpoint in the repository history.  You don't have to commit every change, but you should commit *often,* to avoid losing work.  

1. Review the changed files.
2. Enter a short message describing the changes clearly in **Summary**.  
3. Select **Commit to main**.

Good commit messages:

```text
Complete Week 1 practice
Add HW02 notebook
Correct distance conversion
Update project proposal
```

Bad commit messages:

```text
stuff
changes
final
asdf
```

A commit is recorded locally. It has **not yet been sent to GitHub**.


## 11. Push to GitHub

After committing:

1. Select **Push origin**.
2. Let GitHub Desktop finish the push.

```text
LOCAL REPOSITORY
      ↓ Push
REMOTE REPOSITORY ON GITHUB
```

A push sends committed changes from your computer to GitHub.com.


## 12. Verify the Change on GitHub

Always verify the result.

Open the repository on GitHub.com and confirm:

1. the correct file exists;
2. it is in the correct folder;
3. its filename is correct;
4. the latest changes appear;
5. the latest commit appears.

For an assignment, verify this **before submitting the GitHub link to Canvas**.


### The complete workflow

```text
1. Create repository from template
                    ↓
2. Clone with GitHub Desktop
                    ↓
3. Keep it in one permanent location
                    ↓
4. Open the repository folder in VS Code
                    ↓
5. Edit and save
                    ↓
6. Review in GitHub Desktop
                    ↓
7. Commit
                    ↓
8. Push origin
                    ↓
9. Verify on GitHub.com
```


## 13. Common Mistakes

### Editing the wrong copy
A file in `Downloads` is not automatically the same file as the one inside your repository.

**Work inside the repository folder.**

### Changing required folder names
Do not rename `week02` to `Python Week`.

### Changing required filenames
If the required file is `abc123_hw02.ipynb`, do not submit `My Homework Final.ipynb`.

### Saving without committing
The change exists locally but is not recorded in Git history.

### Committing without pushing
The commit exists locally but GitHub.com still has the older version.

### Not verifying
Always inspect the final version on GitHub.
