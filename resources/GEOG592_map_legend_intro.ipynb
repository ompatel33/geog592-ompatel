```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python


# getting the US census data

url_0 = "https://www2.census.gov/geo/tiger/GENZ2024/shp/" # url for census tiger data 
url_county = "cb_2024_us_county_500k.zip" # county level
url_state = "cb_2024_us_state_500k.zip" # state level 


## Code for county level data 

counties = gpd.read_file(url_0+url_county)
nc_counties = counties.loc[counties["STATEFP"] == "37"].copy()
nc_counties.plot(edgecolor="black", facecolor="white")
```

#### Marking the Projection

```python
def ncplot(ax,stroke,fill):    
    ncc = nc_counties.to_crs(6543).copy() # project to NAD83 / North Carolina (ftUS) for area calculations
    ncc.plot(ax=ax, edgecolor=stroke, facecolor=fill)
```

```python
fig, ax = plt.subplots(figsize=(5,10)) # fig size parameters use (width, height) as parameters.
ncplot(ax, 'green', 'grey')
ax.set_title("North Carolina", fontsize=25, pad=12, fontname="Times New Roman")
ax.text(2036476.502, 831990.2379, "X", ha='center', va='bottom', fontsize='20')
plt.show()
# EXTRA CREDIT!! Code in a loop (rough loop, he doesn't want it to be perfect) assign the x and the y of the centroid and assign the name here.
# for each row in rows: row [x] and row name.
```

#### Extra Credit: Building a Loop

```python
nc_projected = nc_counties.to_crs(epsg=5070)
county_centroids = nc_projected.copy()
county_centroids["geometry"] = nc_projected.geometry.centroid
county_centroids = county_centroids.to_crs(nc_counties.crs)

fig, ax = plt.subplots(figsize=(12, 10))
nc_counties.plot(ax=ax, edgecolor="black", facecolor="white", linewidth=0.6)
county_centroids.plot(ax=ax, color="red", markersize=8)

for _, county in county_centroids.iterrows():
    x, y = county.geometry.x, county.geometry.y
    ax.annotate(
        text=county["NAME"],xy=(x, y),ha="center",va="center",fontsize=6
    )
ax.set_title("North Carolina Counties")
ax.set_axis_off()
plt.show()
```
